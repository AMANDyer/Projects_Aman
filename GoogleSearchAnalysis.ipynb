```python
import pandas as pd
from pytrends.request import TrendReq
import matplotlib.pyplot as plt
from pytrends.exceptions import TooManyRequestsError
import time

trends=TrendReq()
```

```python
## it will be analysisng the google search trends on the queries based on the "Macine LeARNING",
## so lets create a DataFrame of the top 10 countires which search for "ML" on google
```

```python

trends.build_payload(kw_list=["Machine Learning"])
data=trends.interest_by_region()

data=data.sort_values(by="Machine Learning",ascending=False)
data=data.head(10)
print(data)
```

```python
## so according to the above results,the search queires based on "Machine Learning" are mostly done in china
## we can also visulaise this data

```

```python
data.reset_index().plot(x="geoName",
                       y="Machine Learning",
                       figsize=(15,12),kind="bar")
plt.style.use('fivethirtyeight')
plt.show()
```

```python
## so as we all know that ML has been the focus of so  many companies and students for the last 3-4
## years so lets have a look at the trend of searches to see how the total search queries based on "ML"
## incresed or decreased on google:
```

```python
##data=TrendReq(hl='en-US',tz=360)
##data.build_payload(kw_list=["Machine Learning"])
try:
    data=TrendReq(hl='en-US',tz=360)
    data.build_payload(kw_list=["Machine Learning"])
    data=data.interest_over_time()
except TooManyRequestsError:
    time.sleep(60)
    # Optionally retry here
##data=data.interest_over_time()
fig,ax=plt.subplots(figsize=(15,12))
data['Machine Learning'].plot()
plt.style.use('fivethirtyeight')
plt.title('Total Google Searches for Machine Learing',
         fontweight='bold')
plt.xlabel('Year')
plt.ylabel('Total Count')
plt.show()
```
